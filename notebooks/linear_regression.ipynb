```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../data/points.csv")
df.head()
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(df["x"], df["y"], alpha=0.7)
plt.xlabel("x")
plt.ylabel("y")
plt.title("data/points.csv")
plt.show()
```

```python
with open("../models/model_50") as f:
    a, b = map(float, f.read().strip().split(","))

a, b
```

```python
x_line = pd.Series([df["x"].min(), df["x"].max()])
y_line = a * x_line + b

plt.figure(figsize=(8, 6))
plt.scatter(df["x"], df["y"], alpha=0.7, label="data")
plt.plot(x_line, y_line, color="red", label=f"y = {a:.3f}x + {b:.3f}")
plt.xlabel("x")
plt.ylabel("y")
plt.title("Linear regression fit (models/model_50)")
plt.legend()
plt.show()
```
